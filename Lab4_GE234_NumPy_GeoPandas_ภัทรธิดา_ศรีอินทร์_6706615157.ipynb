## 🔹 ตัวอย่างที่ 1: ใช้ NumPy คำนวณข้อมูล NDVI จากภาพ Raster


```python

import numpy as np

# สร้างข้อมูลตัวอย่างสำหรับภาพ NDVI (Normalized Difference Vegetation Index)
nir = np.array([[0.7, 0.8, 0.6], [0.9, 0.5, 0.3], [0.4, 0.7, 0.2]])
red = np.array([[0.3, 0.4, 0.2], [0.5, 0.3, 0.1], [0.2, 0.3, 0.1]])

# คำนวณค่า NDVI
ndvi = (nir - red) / (nir + red)

print("ค่า NDVI:")
print(ndvi)
```


## 🔹 ตัวอย่างที่ 2: ใช้ NumPy คำนวณค่าเฉลี่ยและค่ามากสุดของ NDVI


```python

print(f"ค่าเฉลี่ย NDVI: {np.mean(ndvi):.2f}")
print(f"ค่า NDVI สูงสุด: {np.max(ndvi):.2f}")
print(f"ค่า NDVI ต่ำสุด: {np.min(ndvi):.2f}")
```


## 🔹 ตัวอย่างที่ 3: ใช้ GeoPandas โหลดและวิเคราะห์ข้อมูล Shapefile


```python
import geopandas as gpd

# โหลดข้อมูลชั้นข้อมูลจังหวัดของประเทศไทยจาก Google Drive
gdf = gpd.read_file("/content/drive/MyDrive/Classroom/GE234 Basic Programming/province")

# แสดงข้อมูล 5 แถวแรก
print(gdf.head())
```


## 🔹 ตัวอย่างที่ 4: คำนวณพื้นที่ของแต่ละจังหวัด


```python
# ตรวจสอบค่า CRS (Coordinate Reference System)
print(gdf.crs)

# คำนวณพื้นที่ของแต่ละจังหวัด (หน่วยเป็นตารางเมตร)
gdf["area_sqkm"] = gdf.geometry.area / 1e6

# แสดงพื้นที่จังหวัด 5 อันดับแรกที่ใหญ่ที่สุด
print(gdf.nlargest(5, "area_sqkm")[["PROV_NAMT", "area_sqkm"]])
```


## 🔹 ตัวอย่างที่ 5: ใช้ GeoPandas ทำ Spatial Join


```python
import geopandas as gpd

# โหลดข้อมูลชั้นข้อมูลอำเภอ
gdf_districts = gpd.read_file("/content/drive/MyDrive/Classroom/GE234 Basic Programming/amp")

# ตรวจสอบ CRS ของทั้งสอง GeoDataFrame
print(f"CRS ของ gdf_districts: {gdf_districts.crs}")
print(f"CRS ของ gdf: {gdf.crs}")

# แปลง CRS ของ gdf_districts ให้เป็น CRS เดียวกับ gdf
# gdf มี CRS เป็น EPSG:32647
if gdf_districts.crs != gdf.crs:
    print(f"ทำการแปลง CRS ของ gdf_districts จาก {gdf_districts.crs} เป็น {gdf.crs}")
    gdf_districts = gdf_districts.to_crs(gdf.crs)

# ทำ Spatial Join ระหว่างอำเภอกับจังหวัด
gdf_joined = gpd.sjoin(gdf_districts, gdf, how="inner", predicate="within")

# แสดงตัวอย่างข้อมูลที่เชื่อมโยงกัน
print(gdf_joined.head())
```

**### แบบฝึกหัด 1:**   ใช้ NumPy คำนวณค่า Mean, Max, Min ของค่า NDVI ในอาร์เรย์ที่สร้างขึ้นเอง

```python
import numpy as np

nir = np.array([[0.3, 0.7, 0.2], [0.8, 0.4, 0.5], [0.6, 0.9, 0.2]])
red = np.array([[0.2, 0.5, 0.1], [0.4, 0.2, 0.3], [0.2, 0.5, 0.1]])

# คำนวณค่า NDVI
ndvi = (nir - red) / (nir + red)

print(f"ค่าเฉลี่ย NDVI: {np.mean(ndvi):.2f}")
print(f"ค่า NDVI สูงสุด: {np.max(ndvi):.2f}")
print(f"ค่า NDVI ต่ำสุด: {np.min(ndvi):.2f}")
```

### **แบบฝึกหัด 2:** ใช้ GeoPandas โหลด Shapefile ของจังหวัด และคำนวณพื้นที่ของแต่ละจังหวัด



```python
import geopandas as gpd

gdf = gpd.read_file("/content/drive/MyDrive/Classroom/GE234 Basic Programming/province")

# แสดงข้อมูล 5 แถวแรก
#print(gdf.head())

# ตรวจสอบค่า CRS (Coordinate Reference System)
print(gdf.crs)

# คำนวณพื้นที่ของแต่ละจังหวัด (หน่วยเป็นตารางเมตร)
gdf["Shape_Area"] = gdf.geometry.area / 1e6

# แสดงพื้นที่จังหวัด 5 อันดับแรกที่ใหญ่ที่สุด
print(gdf.nlargest(5, "Shape_Area")[["PROV_NAMT", "Shape_Area"]])
```

## **` แบบฝึกหัด 3:`** ใช้ GeoPandas ทำ Spatial Join ระหว่างข้อมูลจังหวัดและอำเภอ

```python
import geopandas as gpd

# โหลดข้อมูลชั้นข้อมูลอำเภอ (ต้องมั่นใจว่า gdf ซึ่งเป็นข้อมูลจังหวัดได้ถูกโหลดไว้แล้วจากแบบฝึกหัดที่ 2)
gdf_districts = gpd.read_file("/content/drive/MyDrive/Classroom/GE234 Basic Programming/amp", encoding='TIS-620')

# ตรวจสอบ CRS ของทั้งสอง GeoDataFrame
print(f"CRS ของ gdf_districts: {gdf_districts.crs}")
print(f"CRS ของ gdf: {gdf.crs}")

# แปลง CRS ของ gdf_districts ให้เป็น CRS เดียวกับ gdf หากไม่ตรงกัน
# gdf มี CRS เป็น EPSG:32647 (จากตัวอย่างที่ 3 และแบบฝึกหัดที่ 2)
if gdf_districts.crs != gdf.crs:
    print(f"ทำการแปลง CRS ของ gdf_districts จาก {gdf_districts.crs} เป็น {gdf.crs}")
    gdf_districts = gdf_districts.to_crs(gdf.crs)

# ทำ Spatial Join ระหว่างอำเภอกับจังหวัด
# predicate="within" หมายถึง อำเภอจะต้องอยู่ภายในจังหวัด
gdf_joined_exercise3 = gpd.sjoin(gdf_districts, gdf, how="inner", predicate="within")

# แสดงตัวอย่างข้อมูลที่เชื่อมโยงกัน 5 แถวแรก
print(gdf_joined_exercise3.head())
```

## **แบบฝึกหัด 4**: ใช้ NumPy และ GeoPandas ร่วมกันเพื่อหาข้อมูลจังหวัดที่มี NDVI เฉลี่ยสูงสุด

```python
#ข้อมูล NDVI ของแต่ละจังหวัดที่สร้างขึ้นมา
print("แสดงข้อมูลจังหวัดพร้อมค่าเฉลี่ย NDVI:")
display(gdf[['PROV_NAMT', 'avg_ndvi']].head())
```

```python
import numpy as np

# ตรวจสอบว่า gdf ได้ถูกโหลดไว้แล้ว หากยังไม่ได้โหลด ให้โหลดใหม่
# ตัวอย่าง: gdf = gpd.read_file("/content/drive/MyDrive/Classroom/GE234 Basic Programming/province")

# สร้างข้อมูล NDVI เฉลี่ยแบบสุ่มสำหรับแต่ละจังหวัด
# จำนวนค่า NDVI ต้องเท่ากับจำนวนจังหวัดใน gdf
np.random.seed(42) # เพื่อให้ผลลัพธ์สุ่มเหมือนเดิมทุกครั้งที่รัน
num_provinces = len(gdf)
avg_ndvi_values = np.random.uniform(low=0.1, high=0.9, size=num_provinces)

# เพิ่มคอลัมน์ 'avg_ndvi' เข้าไปใน GeoDataFrame ของจังหวัด
gdf['avg_ndvi'] = avg_ndvi_values

# ค้นหาจังหวัดที่มีค่า NDVI เฉลี่ยสูงสุด
province_with_highest_ndvi = gdf.nlargest(1, 'avg_ndvi')

print("จังหวัดที่มี NDVI เฉลี่ยสูงสุดคือ:")
print(province_with_highest_ndvi[['PROV_NAMT', 'avg_ndvi']])
```
