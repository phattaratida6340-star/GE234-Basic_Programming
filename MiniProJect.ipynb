```python
import pandas as pd
import folium
from folium import plugins
from math import radians, cos, sin, asin, sqrt
import ipywidgets as widgets
from IPython.display import display

# --- 1. ฟังก์ชันคำนวณระยะทาง (Haversine Formula) ---
def get_distance(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * asin(sqrt(a))
    km = 6371 * c
    return km

# --- 2. ตั้งค่า API และดึงข้อมูล ---
MAP_KEY = '1ff50fe26556a4796a872cc9759e64f7'
# ขยายขอบเขตให้กว้างขึ้น (SEA) และย้อนหลัง 3 วัน เพื่อให้มีข้อมูลโชว์แน่นอน
SEA_BOUNDS = '92,5,109,25'
URL = f'https://firms.modaps.eosdis.nasa.gov/api/area/csv/{MAP_KEY}/VIIRS_SNPP_NRT/{SEA_BOUNDS}/3'

def run_fire_monitor(my_lat, my_lon):
    try:
        # ดึงข้อมูล
        df = pd.read_csv(URL)

        if len(df) == 0:
            display(widgets.HTML("<b style='color:orange;'>⚠️ ไม่พบจุดความร้อนในระยะเวลา 3 วันที่ผ่านมา</b>"))
            return

        # สร้างแผนที่
        m = folium.Map(location=[my_lat, my_lon], zoom_start=7, control_scale=True)

        # เพิ่ม Google Satellite
        folium.TileLayer(
            tiles='https://mt1.google.com/vt/lyrs=y&x={x}&y={y}&z={z}',
            attr='Google', name='Google Satellite', overlay=False, control=True
        ).add_to(m)

        # ปักหมุดตำแหน่งของเรา
        folium.Marker(
            [my_lat, my_lon],
            icon=folium.Icon(color='blue', icon='home', prefix='fa'),
            tooltip="ตำแหน่งของคุณ"
        ).add_to(m)

        nearest_dist = float('inf')

        # วนลูปปักหมุดจุดความร้อน
        for _, row in df.iterrows():
            dist = get_distance(my_lat, my_lon, row['latitude'], row['longitude'])
            if dist < nearest_dist:
                nearest_dist = dist

            # เลือกสีตามระยะทาง
            marker_color = 'red' if dist < 50 else 'orange'

            folium.CircleMarker(
                location=[row['latitude'], row['longitude']],
                radius=5,
                color=marker_color,
                fill=True,
                fill_opacity=0.6,
                popup=f"ระยะห่าง: {dist:.2f} กม.<br>อุณหภูมิ: {row['bright_ti4']-273.15:.1f} °C",
                tooltip="คลิกดูระยะทาง"
            ).add_to(m)

        # --- 3. ส่วนแจ้งเตือนผลกระทบ (Alert Dashboard) ---
        status = "🟢 ปลอดภัย"
        advice = "ไม่พบจุดความร้อนในระยะประชิด"
        bg_color = "#28a745" # เขียว

        if nearest_dist < 30:
            status = "🔴 อันตราย (ระดับวิกฤต)"
            advice = "พบไฟไหม้ใกล้มาก! ระวังความร้อนและกลุ่มควันหนาแน่น"
            bg_color = "#dc3545" # แดง
        elif nearest_dist < 100:
            status = "🟡 เฝ้าระวัง (ระดับปานกลาง)"
            advice = "พบจุดความร้อนในรัศมี 100 กม. อาจมีปัญหาฝุ่น PM2.5 ตามทิศทางลม"
            bg_color = "#ffc107" # เหลือง

        alert_box = f"""
        <div style="padding:15px; border-radius:10px; background-color:{bg_color}; color:white; font-family:sans-serif; margin-bottom:10px;">
            <h2 style="margin:0;">สถานะ: {status}</h2>
            <p style="margin:5px 0;">พบจุดความร้อนที่ใกล้ที่สุดห่างจากคุณ: <b>{nearest_dist:.2f} กม.</b></p>
            <p style="margin:0;">คำแนะนำ: {advice}</p>
        </div>
        """

        display(widgets.HTML(alert_box))
        folium.LayerControl().add_to(m)
        display(m)
        print(f"พบจุดความร้อนทั้งหมดในภูมิภาค: {len(df)} จุด")

    except Exception as e:
        print(f"เกิดข้อผิดพลาด: {e}")

# ส่วนควบคุมตำแหน่ง (ลองเปลี่ยนพิกัดเพื่อทดสอบระบบ Alert ได้)
widgets.interact(run_fire_monitor,
                 my_lat=widgets.FloatText(value=13.7563, description='ละติจูด:'),
                 my_lon=widgets.FloatText(value=100.5018, description='ลองจิจูด:'))
```
